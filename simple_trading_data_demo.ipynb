````python
# Create data summary visualization
def create_data_summary(session, symbol):
    """Simple data summary chart"""
    
    # Count records
    counts = []
    labels = []
    colors = []
    
    if not session.daily.empty:
        counts.append(len(session.daily))
        labels.append('Daily')
        colors.append(COLORS['primary'])
    
    for interval, data in session.intraday.items():
        if not data.empty:
            counts.append(len(data))
            labels.append(interval)
            colors.append(COLORS['accent'] if '60min' in interval else COLORS['secondary'])
    
    if not counts:
        print("No data to visualize")
        return
    
    # Create bar chart
    fig = go.Figure()
    fig.add_trace(
        go.Bar(
            x=labels,
            y=counts,
            marker_color=colors,
            text=[f'{c:,}' for c in counts],
            textposition='auto',
            textfont=dict(size=14, color='white')
        )
    )
    
    total = sum(counts)
    fig.update_layout(
        title=f"📊 {symbol} Data Portfolio - {total:,} Total Records",
        xaxis_title="Data Type",
        yaxis_title="Record Count",
        height=500,
        font=dict(color=COLORS['text'])
    )
    
    return fig

# Show the summary
fig1 = create_data_summary(session, SYMBOL)
if fig1:
    fig1.show()
````

````python
# Price chart with multiple timeframes
def create_price_chart(session, symbol):
    """Price chart showing different timeframes"""
    
    # Get recent data (last 30 days)
    recent_start = datetime.now() - timedelta(days=30)
    
    datasets = []
    
    # Daily data
    if not session.daily.empty:
        daily_recent = session.daily[session.daily.index >= recent_start]
        if not daily_recent.empty:
            datasets.append(('Daily', daily_recent, COLORS['primary']))
    
    # Intraday data
    for interval, data in session.intraday.items():
        if not data.empty:
            recent_data = data[data.index >= recent_start]
            if not recent_data.empty:
                color = COLORS['accent'] if '60min' in interval else COLORS['secondary']
                datasets.append((interval, recent_data, color))
    
    if not datasets:
        print("No recent data for price chart")
        return
    
    # Create chart
    fig = go.Figure()
    
    for name, data, color in datasets:
        fig.add_trace(
            go.Scatter(
                x=data.index,
                y=data['Close'],
                name=name,
                line=dict(color=color, width=2)
            )
        )
    
    fig.update_layout(
        title=f"📈 {symbol} Price Action - Multiple Timeframes (Last 30 Days)",
        xaxis_title="Date",
        yaxis_title="Price ($)",
        height=600,
        font=dict(color=COLORS['text'])
    )
    
    return fig

# Show price chart
fig2 = create_price_chart(session, SYMBOL)
if fig2:
    fig2.show()
````

## 💾 Step 4: Export Your Data

**Save for external analysis**

````python
# Export to CSV files
print(f"💾 Exporting {SYMBOL} data to CSV files...")

exported = []

# Export daily data
if not session.daily.empty:
    filename = f"{SYMBOL}_daily.csv"
    session.daily.to_csv(filename)
    exported.append(filename)
    print(f"✅ {filename} - {len(session.daily):,} records")

# Export intraday data
for interval, data in session.intraday.items():
    if not data.empty:
        filename = f"{SYMBOL}_{interval}.csv"
        data.to_csv(filename)
        exported.append(filename)
        print(f"✅ {filename} - {len(data):,} records")

print(f"\n🎯 Exported {len(exported)} files for {SYMBOL}")
print(f"📁 Files: {', '.join(exported)}")
print(f"💡 Ready for your trading analysis!")
````

## 🚀 Step 5: Test Other Symbols

**Quick test of other symbols**

````python
# Test multiple symbols quickly
def quick_test(symbol):
    """Quick data test for any symbol"""
    test_session = SimpleSession(symbol)
    daily = test_session.get_daily()
    hourly = test_session.get_intraday('60min')
    minute = test_session.get_intraday('15min')
    
    results = []
    if not daily.empty:
        results.append(f"Daily: {len(daily):,}")
    if not hourly.empty:
        results.append(f"60min: {len(hourly):,}")
    if not minute.empty:
        results.append(f"15min: {len(minute):,}")
    
    return results

# Test popular symbols
test_symbols = ['AAPL', 'TSLA', 'QQQ', 'NVDA', 'MSFT']

print("🧪 Testing popular symbols:")
print("=" * 40)

for symbol in test_symbols:
    try:
        results = quick_test(symbol)
        if results:
            print(f"✅ {symbol:6}: {' | '.join(results)}")
        else:
            print(f"❌ {symbol:6}: No data")
    except Exception as e:
        print(f"❌ {symbol:6}: Error")

print(f"\n🎯 All data readily available - perfect for trading!")
````
